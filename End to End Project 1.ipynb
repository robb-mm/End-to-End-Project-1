```python
# To check if all days are present in a Time Series
def check_missing_days(ts):
    import pandas as pd

    # Example time series
    # dates = pd.date_range("2023-01-01", "2023-01-10").delete([3, 7])  # missing Jan 4 and Jan 8
    # ts = pd.Series(range(len(dates)), index=dates)
    # ts.index.values[0]='2022-03-01'
    # ts[2]='2024-03-10'; ts[91]='2024-07-10'

    # Expected full range
    full_range = pd.date_range(ts.min(), ts.max(), freq="D")
    # full_range = pd.date_range(ts.index.min(), ts.index.max(), freq="D")

    # Check if all days are present
    missing_days = full_range.difference(ts)
    return missing_days

missing_days = check_missing_days(data.copy()['WorkDate'])
if len(missing_days) == 0:
    print("No missing date")
else:
    print(missing_days)
```

```python
import numpy as np
# NurseHrs-to-Patient Ratio
data['NurseHrs'] = round(data['Hrs_RN'] + data['Hrs_LPN'] + data['Hrs_CNA'], 2)
data['NurseHrs-to-Patient Ratio'] = round(data['NurseHrs'] / data['MDScensus'], 2)
data[(data['NurseHrs-to-Patient Ratio']!=np.inf)&(data['NurseHrs-to-Patient Ratio'].notna())]\
[['PROVNUM', 'WorkDate', 'NurseHrs-to-Patient Ratio', 'NurseHrs', 'MDScensus']].sort_values('NurseHrs-to-Patient Ratio', ascending=False)#, na_position='first') #, 'Hrs_RN', 'Hrs_LPN', 'Hrs_CNA'
```

```python
# Total Nurse Hours Per Hospital Per State Per Month
data['WorkMonth'] = data['WorkDate'].dt.to_period('M')#.astype(str)
# data[['WorkMonth', 'WorkDate']]
data.groupby(['PROVNUM', 'STATE', 'WorkMonth'])['NurseHrs'].sum()

# verify sum using filter
# data[(data['PROVNUM']==15012)&(data['STATE']=='AL')&(data['WorkMonth']=='2024-06')]['NurseHrs'].sum()
```

```python
# understand reliance on temporary staff
data['NurseHrs_ctr'] = round(data['Hrs_RN_ctr'] + data['Hrs_LPN_ctr'] + data['Hrs_CNA_ctr'], 2)
data['NurseHrs_emp'] = round(data['Hrs_RN_emp'] + data['Hrs_LPN_emp'] + data['Hrs_CNA_emp'], 2)

data['NurseHrs_ctr Pct'] = round(data['NurseHrs_ctr'] / data['NurseHrs'] * 100, 2)
data['NurseHrs_emp Pct'] = round(data['NurseHrs_emp'] / data['NurseHrs'] * 100, 2)

# data[data['NurseHrs_ctr Pct'] > 50.0][['NurseHrs_ctr Pct', 'NurseHrs', 'NurseHrs_ctr', 'NurseHrs_emp']].sort_values('NurseHrs_ctr Pct', ascending=False) #
data[data['NurseHrs_ctr Pct'] > 50.0][['PROVNUM', 'WorkDate', 'NurseHrs_ctr Pct', 'NurseHrs_emp Pct']].sort_values('NurseHrs_ctr Pct', ascending=False) #
```

### Analytical Questions

```python
# 1. What is the relationship between nurse staffing levels and patient census?
#    Among the nurse hours recorded, wherein the patient census is not zero,
#    the NurseHrs-to-Patient Ratio statistics are as follows:
#       mean value: 3.37 hrs/patient
#       min value:  0 hrs/patient
#       max value: 193.08 hrs/patient
data[(data['NurseHrs-to-Patient Ratio']!=np.inf)&(data['NurseHrs-to-Patient Ratio'].notna())]\
    [['WorkDate', 'NurseHrs-to-Patient Ratio', 'NurseHrs', 'MDScensus']].describe()
    # [['NurseHrs-to-Patient Ratio', 'NurseHrs', 'MDScensus']].sort_values('NurseHrs-to-Patient Ratio', ascending=False)
# data[(data['NurseHrs-to-Patient Ratio']>0.0)&(data['MDScensus']>0)]\
#     [['WorkDate', 'NurseHrs-to-Patient Ratio', 'NurseHrs', 'MDScensus']].describe()
```

```python
# 2. Which hospitals have the highest staffing hours?
#    ISABELLA GERIATRIC CENTER INC at New York City has the highest staffing hours (NurseHrs).
#    Here's the top 10 list.
data.groupby(['PROVNUM','PROVNAME', 'CITY', 'STATE'])['NurseHrs'].sum().sort_values(ascending=False).head(10).reset_index()
```

```python
# 3. What are the average staffing levels by state?
#    Here are the average staffing levels (NurseHrs) by state. NY tops the list.
data.groupby(['STATE'])['NurseHrs'].mean().round(2).sort_values(ascending=False)
```

```python
# 4. Which facilities have the lowest nurse-to-patient coverage?
#    There are 108 facilities that have zero nurse-to-patient coverage between 1 to 91 days, with WHITE RIVER HEALTHCARE at CALICO ROCK, AR and ST CATHERINE OF SIENA at CALDWELL, NJ the worst.
data[(data['NurseHrs-to-Patient Ratio']!=np.inf)&(data['NurseHrs-to-Patient Ratio'].notna())&(np.isclose(data['NurseHrs-to-Patient Ratio'],0.0))]\
[['STATE', 'CITY', 'PROVNAME']].groupby(['STATE','CITY','PROVNAME']).size().reset_index(name='Number of Days').sort_values('Number of Days', ascending=False)
```

```python
# 5. Are certain states relying more on contract nurses?
#    Yes, 49 of the 52 states have relied more on contract nurses for 9 days at least, PA having the most days with 2544 days.
data[data['NurseHrs_ctr Pct'] > 50.0].groupby(['STATE'], as_index=False).size().sort_values('size', ascending=False)#['STATE'].size#
# data['STATE'].nunique()
```
